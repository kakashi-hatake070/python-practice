```python
# wap to check whether a given number is palindrome or or not using wile loop
a = int(input("Enter a number: "))
num_2 = a
rev = 0

while a != 0:
    rev = rev * 10 + a % 10
    a = a // 10

if num_2 == rev:
    print(num_2, "is a palindrome")
else:
    print(num_2, "is not a palindrome")
```

```python
# wap to check whether a given number is armstrong or not
num = int(input("Enter a number: "))

temp = num
sum = 0

while temp > 0:
    digit = temp % 10
    sum = sum + digit ** 3
    temp = temp // 10

if sum == num:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```
